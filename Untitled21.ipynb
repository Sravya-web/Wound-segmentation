```python
import os
import cv2
import numpy as np
from torch.utils.data import Dataset

class WoundDataset(Dataset):
    def __init__(self, image_dir, mask_dir, transform=None):
        self.image_dir = image_dir
        self.mask_dir = mask_dir
        self.transform = transform
        # List files in image_dir and filter for common image extensions
        self.images = [f for f in os.listdir(image_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]

    def __len__(self):
        return len(self.images)

    def __getitem__(self, index):
        img_path = os.path.join(self.image_dir, self.images[index])
        # Use same filename for mask, assuming they are matched
        mask_path = os.path.join(self.mask_dir, self.images[index])

        # Load image and convert from BGR to RGB
        image = cv2.imread(img_path)
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        # Load mask in grayscale
        mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)

        # Ensure mask is scaled to 0-1 if it's 0-255
        if mask is not None:
            mask = np.where(mask > 127, 1.0, 0.0).astype(np.float32)
        else:
            # Fallback if mask is missing or invalid
            mask = np.zeros((image.shape[0], image.shape[1]), dtype=np.float32)

        if self.transform is not None:
            augmentations = self.transform(image=image, mask=mask)
            image = augmentations["image"]
            mask = augmentations["mask"]

        return image, mask

print("WoundDataset class implemented successfully.")
```

```python
import os
import cv2
import numpy as np
from sklearn.model_selection import train_test_split
from torch.utils.data import Dataset, DataLoader

# Define directory paths to ensure they are available in this scope
TRAIN_IMAGE_DIR = '/content/drive/MyDrive/Segmentation/train_images'
TRAIN_MASK_DIR = '/content/drive/MyDrive/Segmentation/train_masks'

# 1. Redefine WoundDataset to accept a specific file list
class WoundDataset(Dataset):
    def __init__(self, image_dir, mask_dir, file_list, transform=None):
        self.image_dir = image_dir
        self.mask_dir = mask_dir
        self.images = file_list
        self.transform = transform

    def __len__(self):
        return len(self.images)

    def __getitem__(self, index):
        img_path = os.path.join(self.image_dir, self.images[index])
        mask_path = os.path.join(self.mask_dir, self.images[index])

        image = cv2.imread(img_path)
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)
        if mask is not None:
            mask = np.where(mask > 127, 1.0, 0.0).astype(np.float32)
        else:
            mask = np.zeros((image.shape[0], image.shape[1]), dtype=np.float32)

        if self.transform is not None:
            augmented = self.transform(image=image, mask=mask)
            image = augmented["image"]
            mask = augmented["mask"]

        return image, mask

# 2. Get all valid image filenames from the training directory
all_filenames = [f for f in os.listdir(TRAIN_IMAGE_DIR) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]

# 3. Split filenames into training (80%) and validation (20%) sets
train_files, val_files = train_test_split(all_filenames, test_size=0.2, random_state=42)

# 4. Create dataset instances
# Note: train_transform and val_transform must be defined in previous cells
train_dataset = WoundDataset(TRAIN_IMAGE_DIR, TRAIN_MASK_DIR, train_files, transform=train_transform)
val_dataset = WoundDataset(TRAIN_IMAGE_DIR, TRAIN_MASK_DIR, val_files, transform=val_transform)

# 5. Define DataLoaders
batch_size = 8
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False)

print(f"Dataset split complete.")
print(f"Training samples: {len(train_dataset)}")
print(f"Validation samples: {len(val_dataset)}")
print(f"DataLoaders created with batch size: {batch_size}")
```

```python
import torch

# 1. Set the number of epochs
NUM_EPOCHS = 20

train_losses = []
val_losses = []

# 2. Create the training loop
for epoch in range(NUM_EPOCHS):
    # 3. Set model to training mode
    model.train()
    epoch_train_loss = 0

    for images, masks in train_loader:
        # Move data to the correct device
        images = images.to(device)
        masks = masks.to(device).unsqueeze(1) # Add channel dimension if necessary

        # 4. Forward pass and weight update
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, masks)
        loss.backward()
        optimizer.step()

        epoch_train_loss += loss.item()

    # 5. Switch to evaluation mode
    model.eval()
    epoch_val_loss = 0

    # 6. Validation loop
    with torch.no_grad():
        for images, masks in val_loader:
            images = images.to(device)
            masks = masks.to(device).unsqueeze(1)

            outputs = model(images)
            loss = criterion(outputs, masks)
            epoch_val_loss += loss.item()

    # Calculate average losses
    avg_train_loss = epoch_train_loss / len(train_loader)
    avg_val_loss = epoch_val_loss / len(val_loader)

    train_losses.append(avg_train_loss)
    val_losses.append(avg_val_loss)

    # 7. Print progress
    print(f"Epoch [{epoch+1}/{NUM_EPOCHS}] - Train Loss: {avg_train_loss:.4f} | Val Loss: {avg_val_loss:.4f}")

print("Training complete.")
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt

def calculate_iou(preds, targets, threshold=0.5, epsilon=1e-7):
    # Apply sigmoid and threshold to convert logits to binary predictions
    preds = torch.sigmoid(preds)
    preds = (preds > threshold).float()
    targets = targets.float()

    intersection = (preds * targets).sum()
    union = preds.sum() + targets.sum() - intersection
    iou = (intersection + epsilon) / (union + epsilon)
    return iou.item()

def calculate_dice(preds, targets, threshold=0.5, epsilon=1e-7):
    # Apply sigmoid and threshold to convert logits to binary predictions
    preds = torch.sigmoid(preds)
    preds = (preds > threshold).float()
    targets = targets.float()

    intersection = (preds * targets).sum()
    dice = (2. * intersection + epsilon) / (preds.sum() + targets.sum() + epsilon)
    return dice.item()

# 1. & 2. Ensure model is in eval mode and use the correct device
loaded_model.eval()

# 3. Initialize variables for metrics tracking
total_val_loss = 0
total_iou = 0
total_dice = 0
num_batches = len(val_loader)

# 4. Inference loop
with torch.no_grad():
    for images, masks in val_loader:
        images = images.to(device)
        masks = masks.to(device).unsqueeze(1)

        # Forward pass
        outputs = loaded_model(images)

        # Loss calculation
        loss = criterion(outputs, masks)
        total_val_loss += loss.item()

        # Metric calculation for current batch
        total_iou += calculate_iou(outputs, masks)
        total_dice += calculate_dice(outputs, masks)

# 5. Calculate and print average metrics
avg_loss = total_val_loss / num_batches
avg_iou = total_iou / num_batches
avg_dice = total_dice / num_batches

print(f"--- Validation Metrics ---")
print(f"Average Validation Loss: {avg_loss:.4f}")
print(f"Average IoU: {avg_iou:.4f}")
print(f"Average Dice Coefficient: {avg_dice:.4f}")

# 6. Qualitative Visualization
# Grab a batch for visualization
images, masks = next(iter(val_loader))
images_dev = images.to(device)
with torch.no_grad():
    outputs = loaded_model(images_dev)
    preds = torch.sigmoid(outputs)
    preds = (preds > 0.5).float().cpu()

num_samples = min(4, images.shape[0])
fig, axes = plt.subplots(num_samples, 3, figsize=(15, 5 * num_samples))

for i in range(num_samples):
    # Denormalize image for visualization
    image_np = images[i].permute(1, 2, 0).numpy()
    mean = np.array([0.485, 0.456, 0.406])
    std = np.array([0.229, 0.224, 0.225])
    image_vis = std * image_np + mean
    image_vis = np.clip(image_vis, 0, 1)

    # Plot Original
    axes[i, 0].imshow(image_vis)
    axes[i, 0].set_title(f"Sample {i+1}: Original")
    axes[i, 0].axis('off')

    # Plot Ground Truth
    axes[i, 1].imshow(masks[i], cmap='gray')
    axes[i, 1].set_title(f"Sample {i+1}: Ground Truth")
    axes[i, 1].axis('off')

    # Plot Prediction
    axes[i, 2].imshow(preds[i, 0], cmap='gray')
    axes[i, 2].set_title(f"Sample {i+1}: Predicted Mask")
    axes[i, 2].axis('off')

plt.tight_layout()
plt.show()
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
import cv2

# 1. Ensure the model is in evaluation mode
loaded_model.eval()

# 2. Grab a batch from the validation loader
images, masks = next(iter(val_loader))
images_dev = images.to(device)

# 3. Get predictions
with torch.no_grad():
    outputs = loaded_model(images_dev)
    preds = torch.sigmoid(outputs)
    preds = (preds > 0.5).float().cpu()

# 4. Visualization settings
num_samples = min(4, images.shape[0])
fig, axes = plt.subplots(num_samples, 2, figsize=(12, 6 * num_samples))

# ImageNet stats for un-normalization
mean = np.array([0.485, 0.456, 0.406])
std = np.array([0.229, 0.224, 0.225])

for i in range(num_samples):
    # Un-normalize image
    img = images[i].permute(1, 2, 0).numpy()
    img = std * img + mean
    img = np.clip(img, 0, 1)
    img_uint8 = (img * 255).astype(np.uint8)

    # Convert masks to uint8
    gt_mask = (masks[i].numpy() * 255).astype(np.uint8)
    pred_mask = (preds[i, 0].numpy() * 255).astype(np.uint8)

    # Create RGB versions of masks for coloring
    # Green for Ground Truth, Red for Prediction
    gt_overlay = np.zeros_like(img_uint8)
    gt_overlay[:, :, 1] = gt_mask  # Green channel

    pred_overlay = np.zeros_like(img_uint8)
    pred_overlay[:, :, 0] = pred_mask # Red channel

    # Blend overlays with original image
    # Areas where both overlap will appear Yellow
    blended = cv2.addWeighted(img_uint8, 0.7, gt_overlay, 0.3, 0)
    blended = cv2.addWeighted(blended, 1.0, pred_overlay, 0.3, 0)

    # Plot side-by-side: Side-by-side for context, then Overlay for comparison
    axes[i, 0].imshow(img)
    axes[i, 0].set_title(f"Sample {i+1}: Original")
    axes[i, 0].axis('off')

    axes[i, 1].imshow(blended)
    axes[i, 1].set_title(f"Sample {i+1}: Overlay (Green=GT, Red=Pred, Yellow=Both)")
    axes[i, 1].axis('off')

plt.tight_layout()
plt.show()
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt

# 1. Ensure the model is in evaluation mode
loaded_model.eval()

# 2. List to store (IoU, image, mask, prediction)
error_analysis_list = []

# 3. & 4. Iterate through val_loader and get predictions
with torch.no_grad():
    for images, masks in val_loader:
        images_dev = images.to(device)
        masks_dev = masks.to(device).unsqueeze(1)

        outputs = loaded_model(images_dev)

        # 5. Process individual samples
        for i in range(images.shape[0]):
            # Calculate individual IoU using the existing function
            # outputs[i:i+1] keeps the batch dimension for the function
            iou_score = calculate_iou(outputs[i:i+1], masks_dev[i:i+1])

            # Get binary prediction
            pred_mask = torch.sigmoid(outputs[i:i+1])
            pred_mask = (pred_mask > 0.5).float().cpu().squeeze().numpy()

            # Store results (CPU tensors/numpy for easier handling later)
            error_analysis_list.append({
                'iou': iou_score,
                'image': images[i].cpu(),
                'mask': masks[i].cpu().numpy(),
                'pred': pred_mask
            })

# 6. Sort by IoU score (ascending) and pick the first 4
error_analysis_list.sort(key=lambda x: x['iou'])
worst_samples = error_analysis_list[:4]

# 7. & 8. Un-normalize and Visualize the 4 worst samples
fig, axes = plt.subplots(4, 3, figsize=(15, 20))

# ImageNet stats for un-normalization
mean = np.array([0.485, 0.456, 0.406])
std = np.array([0.229, 0.224, 0.225])

for idx, sample in enumerate(worst_samples):
    # Un-normalize image
    img_np = sample['image'].permute(1, 2, 0).numpy()
    img_vis = std * img_np + mean
    img_vis = np.clip(img_vis, 0, 1)

    # Plot Original
    axes[idx, 0].imshow(img_vis)
    axes[idx, 0].set_title(f"Worst {idx+1}: Original (IoU: {sample['iou']:.4f})")
    axes[idx, 0].axis('off')

    # Plot Ground Truth
    axes[idx, 1].imshow(sample['mask'], cmap='gray')
    axes[idx, 1].set_title(f"Worst {idx+1}: Ground Truth")
    axes[idx, 1].axis('off')

    # Plot Prediction
    axes[idx, 2].imshow(sample['pred'], cmap='gray')
    axes[idx, 2].set_title(f"Worst {idx+1}: Prediction")
    axes[idx, 2].axis('off')

plt.tight_layout()
plt.show()
```

```python
import segmentation_models_pytorch as smp
import torch

# 1. & 2. & 3. & 4. Instantiate Attention U-Net model
# Using decoder_attention_type='scse' adds Concurrent Spatial and Channel Squeeze & Excitation
attention_model = smp.Unet(
    encoder_name='resnet34',        # ResNet-34 backbone
    encoder_weights='imagenet',    # Pre-trained weights
    decoder_attention_type='scse', # Enable attention mechanism
    in_channels=3,                  # RGB input
    classes=1,                      # Binary segmentation output
)

# 5. Move the model to the active device (defined in previous cells)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
attention_model = attention_model.to(device)

# 6. Print confirmation message
print(f"Attention U-Net model (SCSE) with ResNet-34 backbone initialized and moved to: {device}")
```

```python
from segmentation_models_pytorch.losses import DiceLoss
from torch.optim import Adam

# 1. & 2. Instantiate the DiceLoss function for binary segmentation
attention_criterion = DiceLoss(mode='binary')

# 3. Initialize the Adam optimizer for the attention model
# Using the same learning rate (0.0001) for consistent comparison
attention_optimizer = Adam(attention_model.parameters(), lr=0.0001)

# 4. Print confirmation message
print("Dice Loss function and Adam optimizer re-initialized for the Attention U-Net model.")
```

```python
import torch

# 1. Initialize lists to track losses
attention_train_losses = []
attention_val_losses = []

# 2. Execute training loop for NUM_EPOCHS (same as standard U-Net)
for epoch in range(NUM_EPOCHS):
    # 3. Training phase
    attention_model.train()
    epoch_train_loss = 0

    for images, masks in train_loader:
        images = images.to(device)
        masks = masks.to(device).unsqueeze(1) # Match output shape (B, 1, H, W)

        attention_optimizer.zero_grad()
        outputs = attention_model(images)
        loss = attention_criterion(outputs, masks)
        loss.backward()
        attention_optimizer.step()

        epoch_train_loss += loss.item()

    # 4. Validation phase
    attention_model.eval()
    epoch_val_loss = 0

    with torch.no_grad():
        for images, masks in val_loader:
            images = images.to(device)
            masks = masks.to(device).unsqueeze(1)

            outputs = attention_model(images)
            loss = attention_criterion(outputs, masks)
            epoch_val_loss += loss.item()

    # 5. Calculate and record average losses
    avg_train_loss = epoch_train_loss / len(train_loader)
    avg_val_loss = epoch_val_loss / len(val_loader)

    attention_train_losses.append(avg_train_loss)
    attention_val_losses.append(avg_val_loss)

    print(f"Epoch [{epoch+1}/{NUM_EPOCHS}] - Attention Train Loss: {avg_train_loss:.4f} | Attention Val Loss: {avg_val_loss:.4f}")

# 6. Confirm completion
print("Attention U-Net training process complete.")
```

```python
import matplotlib.pyplot as plt

# 1. Define the range of epochs for the x-axis
epochs_range = range(1, NUM_EPOCHS + 1)

# 2. & 3. Create the figure and plot loss curves for both models
plt.figure(figsize=(12, 7))

# Standard U-Net losses
plt.plot(epochs_range, train_losses, label='Standard U-Net Train Loss', color='blue', linestyle='--', marker='o')
plt.plot(epochs_range, val_losses, label='Standard U-Net Val Loss', color='blue', linestyle='-', marker='x')

# Attention U-Net losses
plt.plot(epochs_range, attention_train_losses, label='Attention U-Net Train Loss', color='red', linestyle='--', marker='o')
plt.plot(epochs_range, attention_val_losses, label='Attention U-Net Val Loss', color='red', linestyle='-', marker='x')

# 4. Add title and labels
plt.title('Convergence Comparison: Standard U-Net vs. Attention U-Net')
plt.xlabel('Epoch')
plt.ylabel('Dice Loss')

# 5. Include a legend
plt.legend()

# Add grid for better readability
plt.grid(True)

# 6. Display the plot
plt.show()
```
